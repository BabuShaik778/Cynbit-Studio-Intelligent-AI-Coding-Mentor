```python
# 2. IMPORT LIBRARIES
import os
import json
import zipfile
import joblib
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
# 4. EXPLORE DATASET
print("\nDataset shape:", df.shape)

print("\nAvailable topics:")
print(df["topic"].value_counts())

print("\nDifficulty distribution:")
print(df["difficulty"].value_counts())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# 5. TEXT PREPROCESSING
df["combined_text"] = (
    df["title"].fillna("") + " " +
    df["topic"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["skills"].fillna("")
)
print("\nText preprocessing completed!")
```

```python
# 6. TRAIN TF-IDF RECOMMENDATION MODEL
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=3000
)

exercise_vectors = vectorizer.fit_transform(
    df["combined_text"]
)

joblib.dump(vectorizer, "vectorizer.joblib")
joblib.dump(exercise_vectors, "exercise_vectors.joblib")

print("\nTF-IDF model trained!")
print("Feature matrix:", exercise_vectors.shape)
```

```python
 # 7. RECOMMENDATION FUNCTION
def recommend_exercises(topic, difficulty=None, top_n=5):

    query_vector = vectorizer.transform([topic])

    scores = cosine_similarity(
        query_vector,
        exercise_vectors
    ).flatten()

    results = df.copy()
    results["similarity_score"] = scores

    if difficulty:

        matching = results[
            results["difficulty"].str.lower()
            == difficulty.lower()
        ]

        if not matching.empty:
            results = matching

    results = results.sort_values(
        "similarity_score",
        ascending=False
    )

    return results.head(top_n)[
        [
            "title",
            "topic",
            "difficulty",
            "description",
            "similarity_score"
        ]
    ]
```

```python
# 8. TEST RECOMMENDATIONS
print("\nHASH MAP RECOMMENDATIONS")

display(
    recommend_exercises(
        "I am weak in dictionary and frequency counting",
        "Easy"
    )
)

print("\nTREE RECOMMENDATIONS")

display(
    recommend_exercises(
        "I want to practice binary tree DFS and BFS",
        "Medium"
    )
)

print("\nDYNAMIC PROGRAMMING RECOMMENDATIONS")

display(
    recommend_exercises(
        "I need to improve Fibonacci and knapsack",
        "Easy"
    )
)
```

```python
# 9. CREATE SAMPLE STUDENT PERFORMANCE DATA
student_data = pd.DataFrame([
    ["Student_01", "Hash Map", 60, 4],
    ["Student_01", "Trees", 45, 5],
    ["Student_01", "Dynamic Programming", 35, 6],
    ["Student_01", "Strings", 80, 2],
    ["Student_01", "Arrays", 75, 3],
    ["Student_01", "Graphs", 40, 5]
], columns=[
    "student", "topic", "score", "attempts"
])

student_data.to_csv(
    "sample_student_performance.csv",
    index=False
)

print("\nSample student performance:")
display(student_data)
```

```python
# 10. IDENTIFY WEAK TOPICS
def identify_weak_topics(data, threshold=60):

    return data[
        data["score"] < threshold
    ].sort_values("score")

weak_topics = identify_weak_topics(student_data)

print("\nTopics requiring improvement:")
display(weak_topics)
```

```python
# 11. PERSONALIZED LEARNING PLAN
def personalized_learning_plan(data):

    weak = identify_weak_topics(data)
    plans = []

    for _, row in weak.iterrows():

        recommendations = recommend_exercises(
            row["topic"],
            difficulty="Easy",
            top_n=2
        )

        for _, exercise in recommendations.iterrows():

            plans.append({
                "student": row["student"],
                "weak_topic": row["topic"],
                "current_score": row["score"],
                "recommended_exercise": exercise["title"],
                "difficulty": exercise["difficulty"]
            })

    return pd.DataFrame(plans)


learning_plan = personalized_learning_plan(student_data)

display(learning_plan)

learning_plan.to_csv(
    "personalized_learning_plan.csv",
    index=False
)
```
